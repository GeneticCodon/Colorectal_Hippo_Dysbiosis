# 01 · TCGA intake & QC
Loads TCGA-COAD counts + clinical via manifest and runs basic QC.

```python
# Read manifests
from pathlib import Path
import pandas as pd, numpy as np
BASE = "/content/drive/MyDrive/Colorectal_Hippo_Dysbiosis"

man = pd.read_csv(f"{BASE}/config/manifest_tcga.csv").set_index("key")["path"].to_dict()

# Load minimal files (paths must exist in your Drive)
counts = pd.read_csv(man["counts"], sep=None, engine="python")
clinical = pd.read_csv(man["clinical"])
print("Counts shape:", counts.shape)
print("Clinical shape:", clinical.shape)

# Basic QC
assert counts.shape[0] > 0 and counts.shape[1] > 0, "Empty counts file"
assert clinical.shape[0] > 0, "Empty clinical file"

# Save sample lists
samples = counts.columns[1:] if counts.columns[0].lower() in ["gene","gene_id","symbol"] else counts.columns
pd.Series(samples, name="sample_id").to_csv(f"{BASE}/results/qc/tcga_sample_list.csv", index=False)
print("Saved sample list to results/qc/")
```
