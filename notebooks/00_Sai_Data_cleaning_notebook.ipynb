```python
# --- Imports & helpers ---
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from pathlib import Path
import json, textwrap, warnings; warnings.filterwarnings("ignore")

def save_table(df, name, subdir="tables"):
    out = f"{BASE}/results/{subdir}/{name}"
    Path(out).parent.mkdir(parents=True, exist_ok=True)
    df.to_csv(out, index=True)
    print("saved ->", out)

def read_manifest(path):
    df = pd.read_csv(path)
    assert set(["key","path"]).issubset(df.columns), "manifest needs columns: key,path"
    return df.set_index("key")["path"].to_dict()

print("Project base:", BASE)
```
